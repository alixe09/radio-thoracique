# EDA — Chest X-Ray Images (Pneumonia)

Exploration du dataset avant entraînement : nombre d'images par classe et par split, aperçu visuel, tailles d'images.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image

DATA_DIR = Path("../data/raw/chest_xray")
```

```python
for split in ["train", "val", "test"]:
    for cls in ["NORMAL", "PNEUMONIA"]:
        folder = DATA_DIR / split / cls
        n = len(list(folder.glob("*.jpeg"))) if folder.exists() else 0
        print(f"{split:6s} / {cls:10s} : {n} images")
```

On observe un fort déséquilibre en faveur de la classe PNEUMONIA dans `train/`, 
et un dossier `val/` beaucoup trop petit (16 images) pour être utilisable tel quel — 
d'où le choix de reconstruire une validation par split stratifié de `train/` 
(voir `src/preprocessing.py`).

```python
fig, axes = plt.subplots(2, 4, figsize=(14, 7))
for row, cls in enumerate(["NORMAL", "PNEUMONIA"]):
    folder = DATA_DIR / "train" / cls
    samples = list(folder.glob("*.jpeg"))[:4]
    for col, img_path in enumerate(samples):
        img = Image.open(img_path)
        axes[row, col].imshow(img, cmap="gray")
        axes[row, col].set_title(f"{cls}\n{img.size}")
        axes[row, col].axis("off")
plt.tight_layout()
plt.show()
```
